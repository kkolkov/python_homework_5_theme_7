Тема Параллельные вычисления. Многопоточность и многопроцессность

```python
import threading
import time
```

Задание 1.
Напишите программу, которая создаёт 2 потока для вычисления квадратов и кубов целых чисел от 1 до 10.

```python
def print_squares():
    for i in range(1, 11):
        print(f"Квадрат {i} = {i ** 2}")
```

```python
def print_cubes():
    for i in range(1, 11):
        print(f"Куб {i} = {i ** 3}")
```

```python
# Создание потоков
t1 = threading.Thread(target=print_squares)
t2 = threading.Thread(target=print_cubes)
```

```python
# Запуск потоков
t1.start()
t2.start()
```

```python
# Ожидание завершения потоков
t1.join()
t2.join()
```

Задание 2.
Напишите программу, которая создаёт несколько потоков для выполнения функции, которая выводит целые числа от 1 до 10 с задержкой в 1 секунду.

```python
def print_numbers(name):
    for i in range(1, 11):
        print(f"{name}: {i}")
        time.sleep(1)
```

```python
# Создаём несколько потоков (например, 3)
threads = []
for n in range(3):
    t = threading.Thread(target=print_numbers, args=(f"Поток {n+1}",))
    threads.append(t)
    t.start()
```

```python
# Ждём завершения всех потоков
for t in threads:
    t.join()
```
